#Задача 1
Написать функцию на вход которой подается строка, состоящая из латинских букв.
Функция должна вернуть количество гласных букв (a, e, i, o, u) в этой строке. Написать тесты для кода

```python
import unittest

def count_vovels(sequence):
    vovels = ["a", "e", "i", "o", "u", "A", "E", "I", "O", "U"]
    count = 0
    for i in range(len(sequence)):
        for vovel in vovels:
            if sequence[i] == vovel:
                count += 1
                break
    return count

class test_vovels_count(unittest.TestCase):
    def test_all_vovels(self):
        res = count_vovels("aeiouAEIOU")
        self.assertEqual(res, 10)

    def test_empty(self):
        res = count_vovels("")
        self.assertEqual(res, 0)

    def test_ascii(self):
        sequence = ""
        for i in range(255):
            sequence += chr(i)
        res = count_vovels(sequence)
        self.assertEqual(res, 10)
```

# Задача 2
Написать функцию на вход, которой подается строка. Функция должна вернуть true, если
каждый символ в строке встречается только 1 раз, иначе должна вернуть false. Написать тесты для кода

```python
import unittest


def are_all_symbols_unique(sequence):
    symbols_set = set()
    for i in range(len(sequence)):
        symbols_set.add(sequence[i])
        if i + 1 > len(symbols_set):
            return False
    return True


class test_symbols_unique(unittest.TestCase):
    def test_ascii(self):
        s = ""
        for i in range(255):
            s += chr(i)
        res = are_all_symbols_unique(s)
        self.assertEqual(res, True)

    def test_empty(self):
        res = are_all_symbols_unique("")
        self.assertEqual(res, True)

    def test_repetition(self):
        s = "abcd11"
        res = are_all_symbols_unique(s)
        self.assertEqual(res, False)
```

# Задача 3
Написать функцию, которая принимает положительное число и возвращает количество
бит равных 1 в этом числе. Написать тесты для кода

```python
import unittest


def count_bin_ones(num):
    if num < 0:  # по условию
        raise ValueError("num must be possitive")
    bin_str = str(bin(num)[2:])
    count = 0
    for i in range(len(bin_str)):
        if bin_str[i] == "1":
            count += 1
    return count


class test_count_bin_ones(unittest.TestCase):
    def test_zero(self):
        res = count_bin_ones(0)
        self.assertEqual(res, 0)

    def test_simple(self):
        res = count_bin_ones(1023)
        self.assertEqual(res, 10)

    def test_negative(self):
        with self.assertRaises(ValueError):
            count_bin_ones(-1)
```

# Задача 4
Написать функцию, которая принимает положительное число. Функция должна вернуть то,
сколько раз необходимо перемножать цифры числа или результат перемножения, чтобы
получилось число состоящее из одной цифры.
Например, для входного числа:
· 39 функция должна вернуть 3, так как 3*9=27 => 2*7=14 => 1*4=4
· 4 функция должна вернуть 0, так как число уже состоит из одной цифры
· 999 функция должна вернуть 4, так как 9*9*9=729 => 7*2*9=126 => 1*2*6=12 =>
1*2=2. Написать тесты для кода

```python
import unittest


def magic(n):
    count = 0
    while n % 10 != n:
        temp = 1
        while n != 0:
            temp = temp * (n % 10)
            n = n // 10
        n = temp
        count += 1
    return count


class test_magic(unittest.TestCase):
    def test_digit(self):
        allpassed = True
        for i in range(10):
            if magic(i) != 0:
                allpassed = False
        self.assertTrue(allpassed)

    def test_number(self):
        res = magic(555)
        self.assertEqual(res, 3)
```
